```python
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from matplotlib.dates import MonthLocator, DateFormatter
import pandas as pd
import numpy as np
from itertools import groupby

forcing = 'FLUXNET-CH4'
mode = 'no_spin_up'

path = f'/share/home/dq076/data/cases/site/{forcing}/{mode}/'

csv_path = f'/share/home/dq076/data/ME/FLUXNET-CH4/FLX_AA-Flx_CH4-META_wetland.csv'
wetland_lists = pd.read_csv(csv_path, header=0)

siteid = wetland_lists['SITE_ID']
# for n in range(len(wetland_lists)):
n = 0
if n == 0:
    siteid = wetland_lists['SITE_ID'][n]
    syear = wetland_lists['YEAR_START'][n]
    eyear = wetland_lists['YEAR_END'][n]
    igbptyp = wetland_lists['IGBP'][n]
    pfttyp = wetland_lists['SITE_CLASSIFICATION'][n]
    unit = r'nmol CH$_4$/m$^2$/s'
    varname = ['f_ch4_ebul_depth']
    color = ["#7aaedc", "#dbb355", "#80328c", "#69aa4c", "#dd5e2d"]
    label = ['ebul']
    dz = [0.01751282, 0.02757897, 0.04547003, 0.07496741, 0.12360037, 
          0.20378255, 0.33598063, 0.55393841, 0.91329003, 1.1369718]
    data_path = f'{path}{siteid}/postdata/{siteid}_hist_{syear}-{eyear}.nc'
    data = xr.open_dataset(data_path)

    # Extract and sum over layers for each variable
    vars_data_list = []
    for k in range(len(varname)):
        # Extract each layer with correct dz[j] for layer j
        layers = [data[varname[k]][:, 0, j] * dz[j] * 1e9 for j in range(10)]
        # Sum over all 10 layers
        summed = np.sum(layers, axis=0)
        vars_data_list.append(summed)

    # Time is the same for all; extract from original data
    time = data['time']
    
    # Compute time interval based on duration (assuming daily data)
    duration_years = len(time) / 365.0
    interval = 12 if duration_years > 5 else 6

    fig = plt.figure(figsize=(12, 9), dpi=300)
    fig.suptitle(f"{n+1} {siteid} {pfttyp} ({igbptyp})", fontsize=16, weight='bold')
    fig.subplots_adjust(left=0.05, right=0.98, 
                        bottom=0.14, top=0.95, hspace=0.3, wspace=0.3)
    gs = GridSpec(3, 4)

    ax = fig.add_subplot(gs[:, :])
    ax.grid(ls="--", lw=0.25, color="#4E616C")
    
    # Plot summed variables
    for k in range(len(varname)):
        ax.plot(time, vars_data_list[k], mfc="white", lw=1, ms=2, 
                color=color[k], label=label[k], alpha=0.85)
    
    # Plot comparison variables
    # ax.plot(time, data['f_ch4_ebul_total'][:, 0] * 1e9, mfc="white", lw=1, ms=2, 
    #         color='black', label='ch4_ebul_tot', alpha=0.85)
    # ax.plot(time, data['f_ch4_surf_ebul'][:, 0] * 1e9, mfc="white", lw=1, ms=2, 
    #         color='red', label='ch4_surf_ebul', alpha=0.85)

    # Print ratio for verification
    print("Ratio (ebul_total / sum_depth):")
    print(data['f_ch4_ebul_total'][:, 0] * 1e9 / vars_data_list[0])

    ax.set_title(f"Total ({unit})")
    ax.legend()
    ax.set_xlabel('Time')
    
    ax.xaxis.set_major_locator(MonthLocator(interval=interval))
    ax.xaxis.set_major_formatter(DateFormatter('%Y-%m'))
    ax.set_xlim(time[0], time[-1])
    ax.tick_params(axis='x', rotation=45)

    plt.tight_layout()
    plt.show()
```

```python
import numpy as np

nl_soil=10
# node depths of each soil layer
z_soi = np.zeros(nl_soil)
for nsl in range(nl_soil):
    z_soi[nsl] = 0.025 * (np.exp(0.5 * (nsl + 1 - 0.5)) - 1.0)
print(z_soi)
# thickness between two soil layer interfaces
dz_soi = np.zeros(nl_soil)
dz_soi[0] = 0.5 * (z_soi[0] + z_soi[1])  # =zi_soi(0)
dz_soi[nl_soil - 1] = z_soi[nl_soil - 1] - z_soi[nl_soil - 2]
for nsl in range(1, nl_soil - 1):
    dz_soi[nsl] = 0.5 * (z_soi[nsl + 1] - z_soi[nsl - 1])
print(dz_soi)
```
