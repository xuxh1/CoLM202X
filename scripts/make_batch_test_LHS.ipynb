# 制作120组随机甲烷参数

## 使用lhs生成随机数

```python
import numpy as np
import pandas as pd
from pyDOE import lhs

# Set the random seed for reproducibility
np.random.seed(0)

# Number of samples
n = 120

# Number of parameters (dimensions)
k = 12

# Generate LHS samples
samples = lhs(k, samples=n)
# Store the 2D samples in a CSV file using pandas
df = pd.DataFrame(samples, columns=[f'param_{i+1}' for i in range(k)])
df.to_csv('/share/home/dq076/mode/ME/251030_r/run/lhs_samples.csv', index=False)

print("Samples saved to 'lhs_samples.csv'")
```

## 生成随机参数

```python
# Read the existing LHS samples CSV (120 rows x 12 columns, values between 0 and 1)
path = '/share/home/dq076/mode/ME/251030_r/run/'
lhs_df = pd.read_csv(f'{path}lhs_samples.csv')

# Read the bounds CSV (3 rows: header, Min, Max; 13 columns including 'Parameter')
bounds_df = pd.read_csv(f'{path}120_colm_samples.csv')  # Adjust filename if different

# Extract min and max values (skip the 'Parameter' column)
mins = bounds_df.iloc[0, 1:].astype(float)  # Min row values
maxs = bounds_df.iloc[1, 1:].astype(float)  # Max row values

# Get parameter names for renaming columns later
param_names = bounds_df.columns[1:].tolist()

# Scale the LHS samples: new_val = (Max - Min) * random + Min
scaled_df = lhs_df.copy()
for i, col in enumerate(lhs_df.columns):
    min_val = mins.iloc[i]
    max_val = maxs.iloc[i]
    scaled_df[col] = (max_val - min_val) * lhs_df[col] + min_val

# Rename columns to match parameter names
scaled_df.columns = param_names

# Save the scaled samples to a new CSV
scaled_df.to_csv(f'{path}scaled_samples.csv', index=False)

print("Scaled samples saved to 'scaled_samples.csv'")
print(scaled_df.head())  # Optional: Preview the first few rows
```
