```python
import os

def rename_files_recursive(root_dir, old_str, new_str, dry_run=True):
    """
    递归搜索指定目录下的所有文件，若文件名中包含 old_str，则替换为 new_str。
    
    参数：
        root_dir (str): 要搜索的根目录路径。
        old_str (str): 要被替换的字符串。
        new_str (str): 替换为的新字符串。
        dry_run (bool): 若为 True，则仅显示将执行的改名操作，不实际修改文件。
    """
    count = 0
    for dirpath, dirnames, filenames in os.walk(root_dir):
        for filename in filenames:
            if old_str in filename:
                old_path = os.path.join(dirpath, filename)
                new_filename = filename.replace(old_str, new_str)
                new_path = os.path.join(dirpath, new_filename)

                if os.path.exists(new_path):
                    print(f"[跳过] 目标文件已存在: {new_path}")
                    continue

                if dry_run:
                    print(f"[预览] 将重命名: {old_path} → {new_path}")
                else:
                    os.rename(old_path, new_path)
                    print(f"[重命名] {old_path} → {new_path}")
                count += 1

    if count == 0:
        print("未找到需要修改的文件。")
    else:
        print(f"\n共{'预览' if dry_run else '完成'}重命名 {count} 个文件。")

if __name__ == "__main__":
    # 示例配置，可根据需要修改
    target_directory = "/share/home/dq076/data/cases/globe/g2"  # 指定路径
    old_pattern = "g1_"                       # 原字符串
    new_pattern = "g2_"                       # 新字符串
    dry_run_mode = False                       # True = 仅预览；False = 实际改名

    rename_files_recursive(target_directory, old_pattern, new_pattern, dry_run=dry_run_mode)
```
